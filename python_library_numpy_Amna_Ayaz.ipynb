# Class Practice: Student Marks Analyzer

Now we will use different NumPy concepts in one small program.

## Problem

We have marks of different students.

We want to find:

- Average marks
- Highest marks
- Lowest marks
- Passed students
- Failed students
- Total passed students
- Total failed students

````python
import numpy as np

marks = [30, 40, 80, 70, 10, 20, 60]

average_marks = (np.mean(marks))
print(average_marks)

highest_marks = (np.max(marks))
print(highest_marks)

lowest_marks = (np.min(marks))
print(lowest_marks)

passed_students = marks[marks >= 50]
print(passed_students)

failed_students = marks[marks < 50]
print(failed_students)
````
